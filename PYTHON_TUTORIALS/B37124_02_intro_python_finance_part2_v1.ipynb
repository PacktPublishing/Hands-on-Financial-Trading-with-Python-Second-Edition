## If and Else Statements
In Python, if and else are used to make decisions in your code. They let your program do one thing if a condition is true and something else if it’s not.

The simplest form of an if statement checks a condition. If the condition is true, the code inside the if block runs:

```python
x = 10
if x > 5:  # Is x greater than 5?
    print("x is greater than 5")  # This runs because the condition is true

# If the condition is false, nothing happens because there's no else yet.
```

```python
# Adding an else
# You can add an else block to handle the case where the condition is false:
x = 3
if x > 5:
    print("x is greater than 5")
else:
    print("x is 5 or less")  # This runs because the condition is false
```

```python
# Using elif (Else If)
# Sometimes you need to check multiple conditions. You can use elif to add more options:
x = 5
if x > 5:
    print("x is greater than 5")
elif x == 5:
    print("x is exactly 5")  # This runs because the condition is true
else:
    print("x is less than 5")
```

### Conditions with `and` and `or`

You can combine multiple conditions using and (all conditions must be true) or or (at least one condition must be true).

**Using and:**

```python
x = 8
if x > 5 and x < 10:  # Is x greater than 5 AND less than 10?
    print("x is between 5 and 10")  # This runs because both conditions are true
else:
    print("x is not between 5 and 10")
```

**Using `or`:**

```python
x = 12
if x < 5 or x > 10:  # Is x less than 5 OR greater than 10?
    print("x is outside the range 5 to 10")  # This runs because x > 10
else:
    print("x is between 5 and 10")
```

Using `and` and `or` with `if-elif-else`

You can combine everything for more complex decision-making:

```python
x = 7
if x < 5:
    print("x is less than 5")
elif x > 5 and x < 10:  # Using 'and'
    print("x is between 5 and 10")
elif x == 10 or x == 15:  # Using 'or'
    print("x is either 10 or 15")
else:
    print("x is greater than 15")
```

**Using `&` and `|` for Logical Operators**

In Python, you can also use & for and and | for or when dealing with boolean values.

**Example with `&`**

```python
# Notice the parentheses!
x = 7
if (x > 5) & (x < 10):  # Both conditions must be true
    print("x is between 5 and 10")
else:
    print("x is not between 5 and 10")
```

**Example with `|`**


```python
x = 12
if (x < 5) | (x > 10):  # At least one condition must be true
    print("x is outside the range 5 to 10")
else:
    print("x is between 5 and 10")
```

When using & (logical AND) and | (logical OR) in Python, you need to use parentheses around each condition to avoid operator precedence issues. Without parentheses, Python might not interpret the conditions the way you intend.

### Nested if statements
You can also nest if statements inside each other for more specific checks:

```python
x = 11
if x > 5:
    if x < 10:
        print("x is between 5 and 10")  # This runs because both conditions are true
    else:
        print("x is greater than or equal to 10")
else:
    print("x is 5 or less")
```

## For Loops
A for loop in Python is used to iterate over a sequence (like a list, dictionary, or range). This means it goes through each item in the sequence one at a time and lets you perform operations on each item.



### Basic Structure of a for Loop

The simplest for loop iterates over a list of items:

```python
stocks = ['AAPL', 'MSFT', 'GOOGL']

for stock in stocks:  # stock takes on each value in the list one at a time
    print(stock)
```

### Iterating Through a List

You can use a for loop to do more than just print each item. For example, let’s print each stock ticker with its index:

```python
for i, stock in enumerate(stocks):  # i is the index, stock is the value
    print(f"Stock {i+1}: {stock}")
```

### Understanding enumerate in Python

enumerate is a built-in Python function that lets you loop through a sequence (like a list, tuple, or string) while keeping track of the index of each item. This is especially useful when you need both the index and the value of each item during iteration.

```python
stocks = ['AAPL', 'MSFT', 'GOOGL']

for index, value in enumerate(stocks):
    print(f"Index: {index}, Value: {value}")
```

- enumerate generates pairs of index and value for each item in the sequence.
- The index starts at 0 by default (like Python indexing).

**Why enumerate Is Useful**

1. Keeps code clean: Without enumerate, you’d have to manually maintain a counter variable to track the index:

```python
stocks = ['AAPL', 'MSFT', 'GOOGL']
index = 0

for stock in stocks:
    print(f"Index: {index}, Value: {stock}")
    index += 1
```

As you can see, enumerate reduces clutter and makes your code easier to read.

2. Saves time and effort: You don’t need to initialize or increment a counter—it’s all handled automatically.

3. Improves debugging: When debugging or logging, it’s often useful to see the index along with the value to pinpoint issues.

4. Flexible starting index: You can specify the starting index using the start parameter:

```python
for index, stock in enumerate(stocks, start=1):  # Start at 1 instead of 0
    print(f"Stock {index}: {stock}")
```

**When to Use enumerate**

1. Whenever you need both index and value: If you’re looping through a sequence and need to know the position of each item, enumerate is the right tool.

2. When working with numbered data: If you’re creating a numbered list or log, enumerate makes this straightforward.

```python
stocks = ['AAPL', 'MSFT', 'GOOGL']
for i, stock in enumerate(stocks, start=1):
    print(f"{i}. {stock}")
```

**Practical Example: Finding an Item by Index**

Let’s say you want to find the position of a specific stock in a list:

```python
stocks = ['AAPL', 'MSFT', 'GOOGL', 'TSLA']

for index, stock in enumerate(stocks):
    if stock == 'GOOGL':
        print(f"GOOGL is at index {index}")
        break  # What is this????? We will come back to this later!
```

### Iterating Through a Dictionary

If you have a dictionary of stock tickers and company names, you can loop through its keys or values:

```python
# Dictionary of stock tickers and their companies
stocks = {
    'AAPL': 'Apple Inc.',
    'MSFT': 'Microsoft Corporation',
    'GOOGL': 'Alphabet Inc.'
}

# Loop through keys (stock tickers)
for ticker in stocks:  # note: you could also do stocks.keys()
    print(ticker)

# Loop through values (company names)
for company in stocks.values():
    print(company)

# Loop through both keys and values
for ticker, company in stocks.items():
    print(f"{ticker}: {company}")
```

### Iterating Through a Nested Dictionary

A nested dictionary is a dictionary where each key’s value is another dictionary. To iterate through it, you can use a for loop and access both the outer and inner dictionary keys and values.

```python
stocks = {
    "AAPL": {
        "Sector": "Technology",
        "Prices": {"Day1": 150.25, "Day2": 151.30, "Day3": 149.80}
    },
    "MSFT": {
        "Sector": "Technology",
        "Prices": {"Day1": 299.50, "Day2": 300.00, "Day3": 298.20}
    },
    "GOOGL": {
        "Sector": "Technology",
        "Prices": {"Day1": 2800.25, "Day2": 2825.00, "Day3": 2790.50}
    }
}
```

```python
for ticker, data in stocks.items():
    print(f"Ticker: {ticker}, Sector: {data['Sector']}")
```

```python
stocks.items()
```

**Accessing Nested Dictionary Values**

To iterate through both levels of the nested dictionary:

```python
for ticker, data in stocks.items():
    print(f"Ticker: {ticker}")
    print(f"  Sector: {data['Sector']}")
    print("  Prices:")
    for day, price in data["Prices"].items():
        print(f"    {day}: ${price}")
```

Breaking Down the Code

*Outer Loop:*

Iterates through stocks.items().
ticker holds the outer key (e.g., "AAPL").
data holds the nested dictionary for each stock.

*Inner Loop:*

Iterates through the "Prices" dictionary using data["Prices"].items().
day holds the day (e.g., "Day1").
price holds the corresponding price (e.g., 150.25).

### Using `range()` to Iterate Through Numbers
The `range()` function generates a sequence of numbers, which is useful in for loops:

Note that unless otherwise specified, range starts at and does NOT include the upper boundary.

```python
# Loop through numbers from 0 to 4
for i in range(5):
    print(i)
```

```python
# Custom range
# Loop through numbers from 3 to 7 (7 is excluded)
for i in range(3, 8):
    print(i)
```

```python
# Loop through numbers with a step of 2
for i in range(0, 10, 2):
    print(i)
```

```python
# Loop through numbers backwards with a step of 2
for i in range(10, 0, -2):
    print(i)
```

```python
# Here is advanced technique
# The range is determined by the length of the iterable
stocks = ['AAPL', 'MSFT', 'GOOGL', 'TSLA']
for i in range(len(stocks)):
  print(stocks[i])
```

### Breaking a Loop When a Condition is Met
You can use the `break` statement to stop the loop early if a condition is met:

```python
stocks = ['AAPL', 'MSFT', 'GOOGL', 'AMZN', 'TSLA']

for stock in stocks:
    print(stock)
    if stock == 'GOOGL':  # Stop the loop when the ticker is 'GOOGL'
        print("Found GOOGL, stopping the loop.")
        break
```

### Summary
1. Lists: Iterate through items directly or use enumerate() for indices.
2. Dictionaries: Use .keys(), .values(), or .items() to loop through different parts.
3. range(): Generate sequences of numbers for controlled iteration.
4. Breaking Loops: Use break to stop a loop early when a condition is met.

## List Comprehensions
A list comprehension is a concise way to create a new list by applying an operation to each item in an existing sequence, optionally filtering items based on a condition. It's powerful because it lets you replace multiple lines of looping code with a single, readable expression.

**Why Are List Comprehensions Important?**
1. Simpler and Cleaner Code:

- They reduce the need for long for loops and temporary variables.
- Code is easier to read and understand.

2. Efficiency:

- List comprehensions are often faster than traditional for loops in Python.

3. Best When:

- You want to create a new list from an existing one.
- You want to filter elements or apply transformations in a single step.

```python
# Basic syntax:
# [expression for item in iterable if condition]
```

Example 1: Transforming Stock Prices

Let’s say you have a list of stock prices, and you want to apply a 10% increase to each price:

```python
prices = [100, 200, 300]

# Traditional for loop
updated_prices = []
for price in prices:
    updated_prices.append(price * 1.1)

print(updated_prices)  # Output: [110.0, 220.0, 330.0]

# Using a list comprehension
print("\nUsing a List Comprehension\n")

updated_prices = [price * 1.1 for price in prices]
print(updated_prices)  # Output: [110.0, 220.0, 330.0]
```

Example 2: Filtering Stock Tickers

Let’s filter a list of stock tickers to include only those starting with A:

Note: While we aren't covering string manipulation in this course, we are showing a couple of cool, easy things to do in Python.
You can find more on strings here:
https://www.w3schools.com/python/python_strings_methods.asp

```python
tickers = ['AAPL', 'MSFT', 'GOOGL', 'AMZN', 'TSLA']

# Traditional for loop
a_stocks = []
for ticker in tickers:
    if ticker.startswith('A'):
        a_stocks.append(ticker)

print(a_stocks)  # Output: ['AAPL', 'AMZN']

# Using a list comprehension
a_stocks = [ticker for ticker in tickers if ticker.startswith('A')]
print(a_stocks)  # Output: ['AAPL', 'AMZN']
```

Example 3: Combining and Filtering

Imagine you have stock prices, and you want to filter out prices below 200 while applying a 5% increase to the remaining prices:

```python
prices = [150, 250, 100, 300]

# List comprehension: Filter and transform
updated_prices = [price * 1.05 for price in prices if price >= 200]
print(updated_prices)  # Output: [262.5, 315.0]
```

Example 4: Extracting Values from a Dictionary

Suppose you have a dictionary of stock tickers and prices, and you want a list of prices above $200:

```python
stock_prices = {'AAPL': 150, 'MSFT': 299, 'GOOGL': 2800, 'TSLA': 450}

# List comprehension: Extract prices above 200
high_prices = [price for price in stock_prices.values() if price > 200]
print(high_prices)
```

Example 5: Nested List Comprehensions

If you want to flatten a list of lists of stock tickers:

```python
nested_tickers = [['AAPL', 'MSFT'], ['GOOGL', 'TSLA'], ['AMZN']]

# Flatten the list
flat_list = [ticker for sublist in nested_tickers for ticker in sublist]
print(flat_list)  # Output: ['AAPL', 'MSFT', 'GOOGL', 'TSLA', 'AMZN']
```

```python
# To help you:
print([sublist for sublist in nested_tickers])
# This is the same thing as just doing:
print(nested_tickers)  # it will just print all elements inside nested_tickers
```

**When to Use List Comprehensions**

1. Quick Transformations: When you need to apply a single operation to a sequence (e.g., applying a formula to prices).
2. Filtering Data: When you need to extract specific items based on a condition (e.g., stocks above a price threshold).
3. Readable Code: When a list comprehension is more concise than a traditional loop without sacrificing clarity.
4. Avoid for Complex Logic: If the operation involves multiple steps or is difficult to understand, stick to regular loops.

## While Loops

A while loop in Python allows you to repeat a block of code as long as a condition is true. It’s different from a for loop, which iterates over a sequence. while loops are used when you don’t know in advance how many times the loop should run.

**Why Use while Loops?**

1. Flexible: Use while loops when the number of iterations depends on a condition that may change during execution.
2. Dynamic Logic: Ideal for problems where the stopping point isn’t predefined, like waiting for user input or processing data until a specific condition is met.
3. Control Flow: Combined with break and continue, while loops give precise control over the flow of execution.

```python
# while condition:
    # Code block to execute repeatedly
```

Example 1: Simple while Loop

Let’s print stock tickers as long as the list isn’t empty:

```python
stocks = ['AAPL', 'MSFT', 'GOOGL']

while stocks:
    print(stocks.pop())  # Removes the last stock from the list
```

**Using `break` to Exit a Loop**

You can use `break` to exit a while loop before the condition becomes false.

Example: breaking when a condition is met

```python
stocks = ['AAPL', 'MSFT', 'GOOGL', 'AMZN']

while stocks:
    stock = stocks.pop(0)  # Removes the first stock
    print(stock)
    if stock == 'MSFT':
        print("Found MSFT, exiting loop.")
        break
```

**Using `continue` to Skip an Iteration**

The continue statement skips the rest of the code in the current iteration and moves to the next one.

Example: Skipping a Specific Stock

```python
stocks = ['AAPL', 'MSFT', 'GOOGL', 'AMZN']

while stocks:
    stock = stocks.pop(0)
    if stock == 'GOOGL':
        print("Skipping GOOGL...")
        continue  # Skip the rest of the loop for this iteration
    print(stock)
```

**Adding a Counter to Prevent Infinite Loops**

A common issue with while loops is infinite loops, where the condition never becomes false. Use counters to avoid this:

```python
i = 0
while i < 5:
    print(f"Iteration {i}")
    i += 1  # Increment the counter
```

**Practical Example: Simulating Stock Prices**

Let’s simulate price changes until a stock hits a target price:

```python
price = 100
target = 120

while price < target:
    print(f"Current price: ${price}")
    price += 5  # Increase the price by $5
print(f"Target reached! Final price: ${price}")
```

### `do-while` Loop Equivalent in Python

Python doesn’t have a built-in `do-while` loop, but you can simulate it using a while loop with a break condition:

Example: always run at least once

```python
stock = None
while True:
    stock = input("Enter a stock ticker: ").upper()
    print(f"You entered: {stock}")
    if stock == 'QUIT':
        print("Exiting...")
        break
```

Here, the loop always runs at least once because the condition is checked after executing the block.

**Understanding while True:**

The statement while True: creates an infinite loop because the condition (True) is always satisfied. This means the loop will continue running forever unless explicitly stopped.

We use it because it ensures the loop will run at least once. But you should always remember the following if using while True:

1. Always Include a break:

- while True: is safe as long as there’s a break to stop the loop when needed.

2. Use for Dynamic Conditions:

- Great for user input, waiting for events, or processing data until a condition is met.

3. Remember the Purpose:

- while True: creates a loop where the stopping logic is written inside the loop, not in the condition itself.
