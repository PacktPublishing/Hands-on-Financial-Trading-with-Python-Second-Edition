### Dictionaries
A dictionary in Python is like a real-world dictionary:

it pairs a "key" with a "value." For example, in a dictionary of stocks, the stock ticker (like AAPL for Apple) can be the key, and the company name (like "Apple Inc.") can be the value.

Think of a dictionary as a container that stores information in pairs, organized like this:

Key: Value

Example:

'MSFT': 'Microsoft Corporation'

Why Use Dictionaries?
Dictionaries are powerful because:

You can quickly look up data by its key.
They are easy to update and manage.
They help you organize related information (like stock tickers and names).

```python
# Creating a dictionary
stocks = {
    "AAPL": "Apple Inc.",
    "MSFT": "Microsoft Corporation",
    "GOOGL": "Alphabet Inc.",
    "AMZN": "Amazon.com Inc."
}

print(stocks)
```

#### Accessing dictionary values from keys

```python
# Accessing a value by its key
print(stocks["AAPL"])
print(stocks["MSFT"])
```

#### Adding new keys or updating existing ones

```python
# Adding a new stock
stocks["TSLA"] = "Tesla Inc."

# Updating an existing stock
stocks["GOOGL"] = "Google LLC"

print(stocks)
```

```python
# Removing a stock
del stocks["AMZN"]
print(stocks)
```

```python
# Check if a key exists
print("AAPL" in stocks)
print("NFLX" in stocks)
```

#### Dictionaries of Lists

Dictionaries of lists are helpful when you:

1. Need to store multiple pieces of data for each key.
2. Want to keep related information organized in one place.
3. Need quick access to all or specific data for each key.


```python
# Dictionary of lists: Stock tickers and their recent prices
stock_prices = {
    "AAPL": [150.25, 151.30, 149.80],  # Prices for Apple Inc.
    "MSFT": [299.50, 300.00, 298.20],  # Prices for Microsoft
    "GOOGL": [2800.25, 2825.00, 2790.50]  # Prices for Alphabet Inc.
}

# Accessing the list of prices for a stock
print(stock_prices["AAPL"])  # Output: [150.25, 151.30, 149.80]

# Accessing a specific price
print(stock_prices["AAPL"][0])  # Output: 150.25
```

```python
# Dictionary of lists: Stock tickers and their sectors
stock_sectors = {
    "AAPL": ["Technology", "Consumer Electronics"],
    "MSFT": ["Technology", "Software"],
    "XOM": ["Energy", "Oil & Gas"]
}

# Accessing the list of sectors for a stock
print(stock_sectors["MSFT"])  # Output: ['Technology', 'Software']

# Accessing the main sector
print(stock_sectors["XOM"][0])  # Output: Energy
```

```python
# Dictionary of lists: Stock tickers and their performance metrics
stock_performance = {
    "AAPL": [10.5, 3.2, 15.7],  # % returns: Year-to-date, 1-year, 5-year
    "MSFT": [8.0, 2.5, 20.1],
    "GOOGL": [12.3, 4.1, 18.9]
}

# Accessing performance data for a stock
print(stock_performance["GOOGL"])

# Accessing a specific performance metric (e.g., 1-year return)
print(stock_performance["MSFT"][1])
```

#### Nested Dictionaries
A nested dictionary is a dictionary where each key has another dictionary as its value. This is useful when you want to store multiple pieces of information for each key in a structured way.

Nested dictionaries are helpful when:

1. You need to organize data hierarchically.
2. Each key needs to store multiple categories of data (like prices, sectors, and performance metrics).
3. You want quick access to specific data points using clear keys.

```python
# Nested dictionary: Stock tickers and their recent prices
stock_prices = {
    "AAPL": {"Day1": 150.25, "Day2": 151.30, "Day3": 149.80},
    "MSFT": {"Day1": 299.50, "Day2": 300.00, "Day3": 298.20},
    "GOOGL": {"Day1": 2800.25, "Day2": 2825.00, "Day3": 2790.50}
}

# Accessing the dictionary for a stock
print(stock_prices["AAPL"])

# Accessing a specific day's price
print(stock_prices["AAPL"]["Day1"])
```

```python
# Nested dictionary: Stock tickers and their sectors
stock_sectors = {
    "AAPL": {"Sector": "Technology", "Sub-sector": "Consumer Electronics"},
    "MSFT": {"Sector": "Technology", "Sub-sector": "Software"},
    "XOM": {"Sector": "Energy", "Sub-sector": "Oil & Gas"}
}

# Accessing the dictionary for a stock
print(stock_sectors["MSFT"])

# Accessing the main sector
print(stock_sectors["XOM"]["Sector"])
```

```python
# Nested dictionary: Stock tickers and their performance metrics
stock_performance = {
    "AAPL": {"YTD": 10.5, "1-Year": 3.2, "5-Year": 15.7},
    "MSFT": {"YTD": 8.0, "1-Year": 2.5, "5-Year": 20.1},
    "GOOGL": {"YTD": 12.3, "1-Year": 4.1, "5-Year": 18.9}
}

# Accessing the dictionary for a stock
print(stock_performance["GOOGL"])  # Output: {'YTD': 12.3, '1-Year': 4.1, '5-Year': 18.9}

# Accessing a specific performance metric (e.g., 1-year return)
print(stock_performance["MSFT"]["1-Year"])  # Output: 2.5
```

```python
# Comprehensive nested dictionary: Stock data
stocks = {
    "AAPL": {
        "Prices": {"Day1": 150.25, "Day2": 151.30, "Day3": 149.80},
        "Sector": {"Main": "Technology", "Sub-sector": "Consumer Electronics"},
        "Performance": {"YTD": 10.5, "1-Year": 3.2, "5-Year": 15.7}
    },
    "MSFT": {
        "Prices": {"Day1": 299.50, "Day2": 300.00, "Day3": 298.20},
        "Sector": {"Main": "Technology", "Sub-sector": "Software"},
        "Performance": {"YTD": 8.0, "1-Year": 2.5, "5-Year": 20.1}
    },
    "GOOGL": {
        "Prices": {"Day1": 2800.25, "Day2": 2825.00, "Day3": 2790.50},
        "Sector": {"Main": "Technology", "Sub-sector": "Internet Services"},
        "Performance": {"YTD": 12.3, "1-Year": 4.1, "5-Year": 18.9}
    }
}

# Accessing all data for AAPL
print(stocks["AAPL"])

# Accessing a specific price for AAPL
print(stocks["AAPL"]["Prices"]["Day1"])  # Output: 150.25

# Accessing the sector for MSFT
print(stocks["MSFT"]["Sector"]["Main"])  # Output: Technology

# Accessing the 1-year return for GOOGL
print(stocks["GOOGL"]["Performance"]["1-Year"])  # Output: 4.1
```

#### Casting to Create a Dictionary
You can use the dict() function to convert data into a dictionary. Let’s use stock tickers and their company names as an example.

Casting is useful when:

1. You already have data in lists, tuples, or other formats.
2. You want to transform the data into a structured format for easier access and management.
3. You need a quick way to build a dictionary without manually defining it.

```python
# List of tuples: (Ticker, Company Name)
stock_data = [("AAPL", "Apple Inc."), ("MSFT", "Microsoft Corporation"), ("GOOGL", "Alphabet Inc.")]

# Casting to dictionary
stocks = dict(stock_data)

print(stocks)
```

```python
# Two separate lists
tickers = ["AAPL", "MSFT", "GOOGL"]
company_names = ["Apple Inc.", "Microsoft Corporation", "Alphabet Inc."]

# Pairing and casting to dictionary
stocks = dict(zip(tickers, company_names))

print(stocks)
```

What does `zip` do?

The `zip()` function in Python takes two or more iterables (like lists, tuples, etc.) and combines them into pairs (or tuples), element by element. It creates a zip object, which is an iterator of these paired elements.

```python
list1 = [1, 2, 3]
list2 = ['a', 'b', 'c']

zipped = zip(list1, list2)
print(list(zipped))
```

**An important note about zip()**

- zip() produces tuples because it’s designed to pair elements generically.

- To turn the pairs into a dictionary, you need to explicitly use dict(zip(...)), as dictionaries require key-value pairs. Without the dict() function, zip() alone has no knowledge of "keys" or "values."

## Summary of Lists, Sets, Tuples, Dictionaries

Here is a brief comparison of all of these data types and when you should use them:

1. **Lists**
- Key Features:

 - Mutable (can be changed after creation).
 - Ordered (elements have a specific sequence).
 - Allows duplicate elements.

- Best for:

 - Storing ordered data where duplicates are allowed.
 - Scenarios requiring frequent additions, deletions, or updates to elements.
 - Iterative tasks, such as looping through a collection of items.

- Example Use Cases:
 - Daily stock prices in a portfolio.
 - A to-do list of tasks.

2. **Sets**
- Key Features:

 - Mutable (but all elements must be immutable).
 - Unordered (elements are not stored in a specific sequence).
 - No duplicate elements allowed.

- Best for:

 - Ensuring uniqueness in a collection.
 - Fast membership tests (element in set).
 - Mathematical operations like unions, intersections, and differences.

- Example Use Cases:

 - A collection of unique tickers in a portfolio.
 - Comparing the tickers in two stock indices to find common or distinct ones.

3. **Tuples**
- Key Features:

 - Immutable (cannot be changed after creation).
 - Ordered (elements have a specific sequence).
 - Allows duplicate elements.

- Best for:
 - Data that should not change during the program’s lifecycle.
 - Fixed collections of related elements that belong together conceptually.

- Example Use Cases:
 - Latitude and longitude coordinates for a stock exchange.
 - Fixed settings like (Ticker, Exchange, Sector).

4. **Dictionaries**
- Key Features:

 - Mutable (can be changed after creation).
 - Unordered (insertion order is preserved as of Python 3.7+).
 - Stores data as key-value pairs.
 - Keys must be unique, values can be duplicated.

- Best for:

 - Mapping relationships between items.
 - Quick lookups by a unique key.
 - Complex structures where each item has multiple attributes.

- Example Use Cases:

 - A mapping of stock tickers to company names (e.g., {'AAPL': 'Apple Inc.'}).
 - Financial metrics for a stock (e.g., {'AAPL': {'Price': 150, 'PE Ratio': 25}}).
